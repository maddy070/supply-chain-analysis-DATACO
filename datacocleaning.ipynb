```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('DataCoSupplyChainDataset.csv', encoding='latin-1')
print(df.shape)
```

```python
# Convert all column names to lowercase and replace spaces with underscores
df.columns = (df.columns
                .str.lower()
                .str.replace(' ', '_')
                )

print(df.columns.tolist())
```

```python
 # removing () too
df.columns = (df.columns
                .str.lower()
                .str.replace(' ', '_')
                .str.replace('(', '')
                .str.replace(')', ''))

print(df.columns.tolist())
```

```python
# Drop columns with no analytical value
df = df.drop(columns=[
    'product_description',  # 100% null
    'product_image',        # just image URLs
    'customer_password',    # sensitive, not needed
    'customer_email',       # sensitive, not needed
    'customer_street',      # too granular
    'customer_zipcode',     # mostly null
    'order_zipcode'         # mostly null
])

print(f'Columns remaining: {df.shape[1]}')
```

```python
# Rename to cleaner names
df = df.rename(columns={
    'order_date_dateorders'    : 'order_date',
    'shipping_date_dateorders' : 'shipping_date'
})
```

```python
# Final shape and null check
print(f'Final shape: {df.shape}')
print(f'\nRemaining nulls:')
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['customer_lname'] = df['customer_lname'].fillna('Unknown')
```

```python
df.to_csv('dataco_cleaned.csv', index=False, encoding='utf-8')
print('Cleaned file saved successfully')
```
